```python
import json
import numpy

def get_mideast_country_dict():
    return {'iran': 'irn',   'israel': 'isr',               'egypt': 'egy',   'saudi arabia': 'sau',
            'turkey': 'tur', 'iraq': 'irq',                 'yemen': 'yem',   'syria': 'syr',
            'jordan': 'jor', 'united arab emirates': 'are', 'lebanon': 'lbn', 'oman': 'omn',
            'kuwait': 'kwt', 'qatar': 'qat',                'bahrain': 'bhr', 'cyprus': 'cyp',
            'palestine': 'pse', }

mideast_country_dict = get_mideast_country_dict()

with open("/ThinkTank-ME/POLECAT-FOR-ME/4_routing_dataset/country_select_max100_min20/country_forecast_results_test.json", "r", encoding='latin-1') as f:
    all_json = json.load(f)

file_train = "../data/country_select_result.json"
beam_size = 1

with open(file_train, "r", encoding='latin-1') as f:
    train_json = json.load(f)

all_number_c = {}
right_number_c = {}
for index_2i, content_2i in mideast_country_dict.items():
    all_number_c[index_2i] = 0
    right_number_c[index_2i] = 0

right_number_gt = 0
all_number_gt = 0

right_number_country = 0
all_number_country = 0

for (index_1i, content_1i), (index_2i, content_2i) in zip(train_json.items(), all_json.items()):
    assert content_1i['input'].split("\n\n[Current Event to Predict]:")[0] == content_2i['input'].split("\n\n[Current Event to Predict]:")[0]

    country_i = content_2i["country_name"]

    for index_3ii, content_3ii in mideast_country_dict.items():
        if country_i == index_3ii:
            all_number_c[index_3ii] = all_number_c[index_3ii] + 1

    if content_2i["correct_forecast"] == []:

        target_i = content_1i["target"]
        output_i = content_1i["output"]
        all_number_country = all_number_country + 1

        for index_1ii, content_1ii in enumerate(output_i):
            if target_i == content_1ii.strip().split("<|")[0]:
                right_number_country = right_number_country + 1
                break
            if index_1ii == beam_size - 1:
                break
    else:
        target_i = content_2i["correct_forecast"]
        output_i = content_1i["output"]
        all_number_gt = all_number_gt + 1

        for index_1ii, content_1ii in enumerate(output_i):
            if content_1ii.strip().split("<|")[0].lower() in target_i:
                right_number_gt = right_number_gt + 1

                for index_3ii, content_3ii in mideast_country_dict.items():
                    if country_i == index_3ii:
                        right_number_c[index_3ii] = right_number_c[index_3ii] + 1

                break
            if index_1ii == beam_size - 1:
                break

print("Training:")
print(f"Micro_average: {right_number_gt} / {all_number_gt + all_number_country} = {right_number_gt / (all_number_gt + all_number_country)}")
print()

Macro_average = []
for index_4i, content_4i in mideast_country_dict.items():
    Macro_average.append(right_number_c[index_4i]/all_number_c[index_4i])
    print(f"{index_4i}: {right_number_c[index_4i]/all_number_c[index_4i]} ({right_number_c[index_4i]}/{all_number_c[index_4i]})")
    
print(Macro_average)
Macro_average = numpy.mean(Macro_average)
print(f"Macro_average: {Macro_average}")
```

```python
import pandas as pd
import json
```

```python
df = pd.read_csv('../MidEast.csv', keep_default_na=False)

news_article = {}
for index, row in df.iterrows():
    subject_i = row["Actor Name"]
    relation_i = row["Event Type"]
    relation_mode_i = row["Event Mode"]
    object_i = row["Recipient Name"]

    c_i = row["Contexts"]

    s_p = row["Story People"]
    s_o = row["Story Organizations"]
    s_l = row["Story Locations"]

    date_i = row["Event Date"]
    id_i = row["Event ID"]
    id_news_i = id_i.split("_")[0]

    if relation_mode_i != "":
        relation_i = f"{relation_i}_{relation_mode_i}"

    if id_news_i not in news_article.keys():
        news_article[id_news_i] = {}
    news_article[id_news_i][id_i] = {"Subject": subject_i, "Relation": relation_i,"Object":object_i, "Contexts": c_i,
                                     "Date":date_i,"Story People":s_p,"Story Organizations": s_o,"Story Locations":s_l}
    
with open("./data_mideast/news_article_orignal.json", mode="w", encoding='latin-1') as f:
    json.dump(news_article, f, indent=4)
```

```python
import json


with open("./data_mideast/news_article_orignal.json", mode="r", encoding='latin-1') as f:
    new_article_o = json.load(f)

new_article_r = {}

for index_i, item_i in new_article_o.items():
    data_i = []
    sp_list = []
    so_list = []
    sl_list = []
    c_list = []

    event_list = []

    for number_ii, (index_ii, item_ii) in enumerate(item_i.items()):
        data_i.append(item_ii["Date"])

        if item_ii["Story People"] != "":
            sp_list.append(item_ii["Story People"])
        else:
            sp_list.append("None")
        if item_ii["Story Organizations"] != "":
            so_list.append(item_ii["Story Organizations"])
        else:
            so_list.append("None")
        if item_ii["Story Locations"] != "":
            sl_list.append(item_ii["Story Locations"])
        else:
            sl_list.append("None")
        if item_ii["Contexts"] != "":
            c_list.append(item_ii["Contexts"])
        else:
            c_list.append("None")

        sss = item_ii["Subject"]
        rrr = item_ii["Relation"]
        ooo = item_ii["Object"]

        event_sentence = get_sentence(sss, ooo, rrr)

        event_list.append(f"{number_ii+1}. {event_sentence}")

    # for i in data_i:
    #     assert i == data_i[0]
    # for i in sp_list:
    #     assert i == sp_list[0]
    # for i in so_list:
    #     assert i == so_list[0]
    # for i in sl_list:
    #     assert i == sl_list[0]

    news_content_i = get_news_content(event_list, sp_list[0], so_list[0], sl_list[0], c_list[0])
    news_date_i = data_i[0]

    new_article_r[index_i] = {"news_article": news_content_i, "news_date": news_date_i}

with open("./data_mideast/news_article_final.json", mode="w", encoding='latin-1') as f:
    json.dump(new_article_r, f, indent=4)
```

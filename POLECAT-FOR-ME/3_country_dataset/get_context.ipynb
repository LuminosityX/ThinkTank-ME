```python
import pandas as pd
```

```python
df_raw = pd.read_csv('/ThinkTank-ME/POLECAT-FOR-ME/2_historical_event/2_3_data_cleaning/data_mideast_max100_min20/MidEast_Raw.csv', sep=',',  keep_default_na=False)
```

```python
df_group_ce = list(df_raw.groupby(["ce_id"]))
len(df_group_ce)
```

```python
max_id_ce = df_raw["ce_id"].max()
max_id_ce
```

```python
from tqdm import tqdm
import json

ce_date = {}

for ce_num in tqdm(range(max_id_ce+1)):
    df_ce_i = df_raw[df_raw["ce_id"] == ce_num]
    time_list = sorted(list(df_ce_i["Date"].unique()))

    
    context_i = {}
    context_i_pre = []
    for time_i in time_list:
        df_ce_time_i = df_ce_i[df_ce_i["Date"]==time_i]
        
        event_i_list = []
        for _, event_i in df_ce_time_i.iterrows():
            s_i = event_i["Subject"]
            r_i = event_i["Relation"]
            o_i = event_i["Object"]

            event_i_str = f"[{s_i}, {r_i}, {o_i}]"
            event_i_list.append(event_i_str)

        ce_date_i = f"Set of events on {time_i} = {{" + ", ".join(event_i_list) +"}\n"
        context_i_pre.append(ce_date_i)
        
        context_i[time_i] = "".join(context_i_pre[:-1])
    
    ce_date[ce_num] = context_i

len(ce_date)
```

```python
with open("./context_mideast_max100_min20.json", mode="w", encoding='latin-1') as f:
    json.dump(ce_date, f, indent=4)
```

1. First get the all object of test datasets.

```python
import json

with open(f"./country_max100_min20/mideast_test_object.json", mode="r", encoding='latin-1') as f:
    all_test_dataset = json.load(f)

test_object = []

for index_i, content_i in all_test_dataset.items():
    test_object.append(content_i["target"])

test_object = list(set(test_object))

with open(f"./country_max100_min20/mideast_test_object_list.json", mode="w", encoding='latin-1') as f:
    json.dump(test_object, f, indent=4)
```

2. Use the tokenizer (llama3.1) to get the id version of test object.

```python
from transformers import AutoModel, AutoTokenizer
import torch
import json
import numpy as np

tokenizer = AutoTokenizer.from_pretrained('meta/Llama-3.1-8B-Instruct', trust_remote_code=True)

# print(tokenizer.eos_token_id)  # 128009
# print(tokenizer.sep_token_id)  # None
# print(tokenizer.pad_token_id)  # None

# eos_token_id
# print(tokenizer.decode(128009))
# <|end_of_text|>
# print(tokenizer.decode(128001))
# print(tokenizer.decode(2969))
# print(tokenizer.decode(72))
# print(tokenizer.decode(7351))

# training example id
# target_id: tensor([[   271,     39,   2969,     72,   7351, 128009]])

print(tokenizer.decode(92995))

with open("./country_max100_min20/mideast_test_object_list.json", mode="r") as f:
    term_rep_json = json.load(f)

id_term_rep = []
for term_rep_i in term_rep_json:

    id_term_rep_i = [271]
    tokenizer_id_i = tokenizer.encode(term_rep_i, add_special_tokens=False)
    id_term_rep_i.extend(tokenizer_id_i)
    # add end_id
    id_term_rep_i.append(128009)

    id_term_rep.append(id_term_rep_i)

print(len(id_term_rep))
print(id_term_rep)
id_term_rep = np.array(id_term_rep, dtype=object)
print(id_term_rep)

np.save("./country_max100_min20/id_term_rep_mideast_test_object_list.npy", id_term_rep)
```
